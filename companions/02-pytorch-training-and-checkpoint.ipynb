# A PyTorch training loop with checkpoint verification

A companion to courses 04–05: train a small nonlinear classifier, select an epoch using validation loss, and reload the chosen state before final evaluation.

**Origin:** new study example created with AI assistance during repository curation; separate from the IBM assignments. Data are synthetic two-moons points generated locally. CPU only, fixed seed, no dataset or checkpoint download.

[Setup](../docs/SETUP.md) · [Course 04 notes](../courses/04-pytorch-foundations/README.md)

## 1. Split before preprocessing

60% training, 20% validation and 20% test. Standardization learns from training data only. The model uses two input features, two hidden layers, and two output logits.

```python
import copy
import io
import json
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.datasets import make_moons
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

SEED = 42
torch.manual_seed(SEED)
torch.set_num_threads(1)
torch.use_deterministic_algorithms(True)
X, y = make_moons(n_samples=2000, noise=0.25, random_state=SEED)
train_idx, remaining_idx = train_test_split(
    np.arange(len(y)), test_size=0.4, stratify=y, random_state=SEED
)
val_idx, test_idx = train_test_split(
    remaining_idx, test_size=0.5, stratify=y[remaining_idx], random_state=SEED
)
assert not (set(train_idx) & set(val_idx) or set(train_idx) & set(test_idx) or set(val_idx) & set(test_idx))
scaler = StandardScaler().fit(X[train_idx])
def tensors(indices):
    return (torch.tensor(scaler.transform(X[indices]), dtype=torch.float32),
            torch.tensor(y[indices], dtype=torch.long))
X_train, y_train = tensors(train_idx)
X_val, y_val = tensors(val_idx)
X_test, y_test = tensors(test_idx)
loader = DataLoader(
    TensorDataset(X_train, y_train), batch_size=64, shuffle=True,
    generator=torch.Generator().manual_seed(SEED)
)
def new_model():
    return nn.Sequential(nn.Linear(2, 32), nn.ReLU(), nn.Linear(32, 32), nn.ReLU(), nn.Linear(32, 2))
model = new_model()
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=1e-4)
print(f"Train / validation / test: {len(train_idx)} / {len(val_idx)} / {len(test_idx)}")
```

## 2. Train and retain the best validation checkpoint

CrossEntropyLoss consumes logits and integer class labels. Each batch clears accumulated gradients. Evaluation combines eval() with inference_mode(): the two operations serve different purposes.

```python
history = []
best_val_loss = float("inf")
best_state = None
best_epoch = None
for epoch in range(1, 81):
    model.train()
    train_loss = 0.0
    for features, labels in loader:
        optimizer.zero_grad(set_to_none=True)
        loss = loss_fn(model(features), labels)
        assert torch.isfinite(loss)
        loss.backward()
        optimizer.step()
        train_loss += loss.item() * len(labels)
    model.eval()
    with torch.inference_mode():
        val_loss = loss_fn(model(X_val), y_val).item()
    history.append((epoch, train_loss / len(train_idx), val_loss))
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = copy.deepcopy(model.state_dict())
        best_epoch = epoch
assert best_state is not None
model.load_state_dict(best_state)
model.eval()
print(f"Selected epoch: {best_epoch}; validation loss: {best_val_loss:.4f}")
```

## 3. Reload the checkpoint and evaluate once

An in-memory buffer exercises the same serialization/reload path without leaving model binaries in the repository. Load only a state dictionary with weights_only=True. Compare logits from the original and reloaded models before reading the held-out score.

```python
checkpoint = io.BytesIO()
torch.save(model.state_dict(), checkpoint)
checkpoint.seek(0)
reloaded = new_model()
reloaded.load_state_dict(torch.load(checkpoint, map_location="cpu", weights_only=True))
reloaded.eval()
with torch.inference_mode():
    expected_logits = model(X_test)
    actual_logits = reloaded(X_test)
    assert torch.equal(expected_logits, actual_logits)
    prediction = actual_logits.argmax(dim=1).numpy()
majority_class = int(torch.bincount(y_train).argmax())
metrics = {
    "dataset": "synthetic-two-moons",
    "seed": SEED,
    "training_examples": len(train_idx),
    "validation_examples": len(val_idx),
    "test_examples": len(test_idx),
    "selected_epoch": best_epoch,
    "validation_loss": best_val_loss,
    "test_accuracy": float(accuracy_score(y_test.numpy(), prediction)),
    "test_macro_f1": float(f1_score(y_test.numpy(), prediction, average="macro")),
    "dummy_test_accuracy": float(np.mean(y_test.numpy() == majority_class)),
    "checkpoint_reload_exact": True,
    "device": "cpu",
}
assert metrics["test_accuracy"] > 0.8  # Regression floor for this fixed synthetic example.
print(json.dumps(metrics, indent=2))
```

```python
history_array = np.asarray(history)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(history_array[:, 0], history_array[:, 1], label="Train loss")
ax.plot(history_array[:, 0], history_array[:, 2], label="Validation loss")
ax.axvline(best_epoch, color="black", linestyle="--", alpha=0.5, label="Selected epoch")
ax.set(xlabel="Epoch", ylabel="Cross-entropy loss", title="Training and validation curves")
ax.legend()
fig.tight_layout()
plt.show()
```
